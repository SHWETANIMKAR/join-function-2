```python
import pandas as pd 
```

```python
movies = pd.DataFrame({
    "Genre": ["Drama", "Action", "Comedy"]
}, index=["Dangal", "Jawan", "3 Idiots"])
print(movies)
```

```python
ratings = pd.DataFrame({
    "Rating": [8.4, 7.0, 8.4]
}, index=["Dangal", "Jawan", "3 Idiots"])

print(ratings)
```

```python
#Genre and rating ko join krna hain index ke acording 
```

```python
#LEFT JOIN
print(movies.join(ratings,how='left'))
```

```python
#rigth join
print(movies.join(ratings,how='right'))
```

```python
print(movies.join(ratings,how='inner'))
```

```python
#rigth=rigth table index
#inner=mathching index dega
#outer= dono table ke sare index
```

```python
print(movies.join(ratings,how='outer'))
```
